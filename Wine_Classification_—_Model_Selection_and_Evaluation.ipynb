### Why is feature scaling important?

Differences in feature scale can affect some machine learning algorithms because features with larger numerical values may have a greater influence on the model.

For example, suppose we have two features:

- Age: values between 18 and 80
- Salary: values between 20,000 and 100,000

If an algorithm calculates distances between observations, the salary feature will dominate because its values are much larger than the age values.

For two observations, the Euclidean distance is:

\[
d = \sqrt{(x_1-x_2)^2 + (y_1-y_2)^2}
\]

If the difference in age is 10 but the difference in salary is 20,000, the salary difference will contribute much more to the final distance.

This is especially important for algorithms such as:

- **K-Nearest Neighbors (KNN)**, because it directly uses distances between observations.
- **Support Vector Machines (SVM)**, because the position of observations and the margin between classes depend on feature values.
- **Logistic Regression**, especially when regularization or gradient-based optimization is used.
- **Neural Networks**, because features with very different scales can make optimization more difficult.

A common solution is **feature standardization**:

\[
x' = \frac{x-\mu}{\sigma}
\]

where:

- \(x\) is the original value
- \(\mu\) is the mean of the feature
- \(\sigma\) is the standard deviation

After standardization, features usually have a mean close to 0 and a standard deviation close to 1.

Tree-based algorithms such as **Decision Trees** and **Random Forests** usually do not require feature scaling, because they split the data using conditions such as:

\[
x < \text{threshold}
\]

rather than calculating distances between observations.

**In summary:** feature scaling is especially important for algorithms that depend on distances, feature magnitudes, or numerical optimization.

## Feature Scaling Strategy

Feature scaling is an important preprocessing step because the Wine dataset contains numerical features with substantially different scales.

### Scaling Requirements

| Algorithm | Scaling Strategy |
|---|---|
| Logistic Regression | StandardScaler recommended |
| K-Nearest Neighbors | StandardScaler strongly recommended |
| Gaussian Naive Bayes | No scaling required |
| Linear SVM | StandardScaler strongly recommended |
| RBF SVM | StandardScaler strongly recommended |
| Decision Tree | No scaling required |
| Random Forest | No scaling required |

Distance-based algorithms, such as KNN and RBF SVM, are particularly sensitive to feature scales. Without scaling, features with larger numerical ranges may dominate distance calculations.

Logistic Regression and Linear SVM also benefit from scaling because of their optimization procedures and regularization.

Decision Trees and Random Forests generally do not require feature scaling because their decisions are based on feature thresholds rather than distances.

Gaussian Naive Bayes does not generally require scaling because it models the distribution of each feature separately.

### Selected Scaling Method

StandardScaler will be used for Logistic Regression, KNN, Linear SVM, and RBF SVM.

StandardScaler transforms each feature by subtracting its training-set mean and dividing by its training-set standard deviation.

MinMaxScaler is an alternative that maps the training values of each nonconstant feature to the range [0, 1].

### Data Leakage Prevention

To prevent data leakage, scaling parameters must be estimated using only the training data.

Scikit-learn Pipelines will be used to combine preprocessing and classification. During cross-validation, the scaler will be fitted independently within each training fold, ensuring that validation observations do not influence preprocessing.

The test set will remain unseen throughout model selection and hyperparameter tuning and will only be used for final evaluation.

````python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
````

````python
model_logistic_regression = LogisticRegression(random_state=42)
model_knn = KNeighborsClassifier()
model_gaussian_naive_bayes = GaussianNB()
model_linear_svm = SVC(kernel="linear", random_state=42)
model_rbf_svm = SVC(kernel="rbf", random_state=42)
model_decision_tree = DecisionTreeClassifier(random_state=42)
model_random_forest = RandomForestClassifier(random_state=42)
````

````python
pipe_logistic_regression = Pipeline([
    ("scaler", StandardScaler()),
    ("model", model_logistic_regression)
])

pipe_knn = Pipeline([
    ("scaler", StandardScaler()),
    ("model", model_knn)
])
pipe_gaussian_naive_bayes = Pipeline([
    ("model", model_gaussian_naive_bayes)
])
pipe_linear_svm = Pipeline([
    ("scaler", StandardScaler()),
    ("model", model_linear_svm)
])
pipe_rbf_svm = Pipeline([
    ("scaler", StandardScaler()),
    ("model", model_rbf_svm)
])
pipe_decision_tree = Pipeline([
    ("model", model_decision_tree)
])
pipe_random_forest = Pipeline([
    ("model", model_random_forest)
])
````

````python
models = {
    "Logistic Regression": pipe_logistic_regression,
    "KNN": pipe_knn,
    "Gaussian Naive Bayes": pipe_gaussian_naive_bayes,
    "Linear SVM": pipe_linear_svm,
    "RBF SVM": pipe_rbf_svm,
    "Decision Tree": pipe_decision_tree,
    "Random Forest": pipe_random_forest
}
````

````python
for name, pipeline in models.items():
    print(name)
    print(pipeline)
    print("-" * 50)
````

### Objective

Evaluate the final classifier using class-specific performance metrics, including Precision, Recall, and F1-score.

### Methodology

A classification report is generated using `classification_report()` from `sklearn.metrics`.

The report provides the following metrics for each wine class:

**Precision**

The proportion of samples predicted to belong to a particular class that actually belong to that class.

\[
Precision = \frac{TP}{TP+FP}
\]

**Recall**

The proportion of actual samples belonging to a particular class that the classifier correctly identifies.

\[
Recall = \frac{TP}{TP+FN}
\]

**F1-score**

The harmonic mean of Precision and Recall.

\[
F1 = \frac{2 \cdot Precision \cdot Recall}{Precision+Recall}
\]

**Support**

The number of actual test observations belonging to each class.

### Averaging Methods

The report also includes:

- **Macro Average:** The unweighted mean of each metric across the three classes.
- **Weighted Average:** The mean of each metric weighted by the number of observations in each class.

### Interpretation

Class-specific metrics provide insight into the classifier's strengths and weaknesses.

High Precision and Recall across all three classes would indicate that the classifier performs consistently, rather than achieving high accuracy primarily through one particular class.

## Task 24 — Feature Importance Analysis Using Permutation Importance

### Objective

Investigate which input features have the greatest influence on the predictive performance of the optimized RBF SVM classifier.

### Background

Unlike linear models, an RBF SVM does not provide a straightforward set of feature coefficients that can be directly interpreted as feature importance.

Permutation importance provides a model-agnostic approach for estimating how strongly a trained classifier relies on individual input features.

### Methodology

Permutation importance is calculated using `permutation_importance()` from `sklearn.inspection`.

The procedure consists of the following steps:

1. Evaluate the trained model using the original feature values.
2. Randomly shuffle the values of one feature while keeping the remaining features unchanged.
3. Recalculate the model's Macro F1-score.
4. Measure the change in predictive performance.
5. Repeat the procedure multiple times for each feature.

The mean decrease in Macro F1-score is used as an estimate of feature importance.

In this project, permutation importance is computed on the training set as an exploratory analysis, leaving the held-out test set reserved for final performance evaluation.

### Interpretation

Features associated with a larger positive decrease in model performance are considered more influential for the trained classifier.

However, permutation importance measures model reliance rather than causal relationships.

Additionally, strongly correlated features may share predictive information, which can affect their individual importance scores.

### Expected Outcome

A ranked table and a horizontal bar chart illustrating the relative importance of the 13 chemical features used to classify wine samples.

````python
from sklearn.inspection import permutation_importance
import matplotlib.pyplot as plt
import pandas as pd

# Calculate permutation importance
perm_result = permutation_importance(
    estimator=final_model,
    X=X_train,
    y=y_train,
    scoring="f1_macro",
    n_repeats=20,
    random_state=42,
    n_jobs=-1
)

# Create feature importance DataFrame
importance_df = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance Mean": perm_result.importances_mean,
    "Importance Std": perm_result.importances_std
})

importance_df = importance_df.sort_values(
    by="Importance Mean",
    ascending=False
)

display(importance_df)

# Plot feature importance
plt.figure(figsize=(10, 6))

plt.barh(
    importance_df["Feature"],
    importance_df["Importance Mean"],
    xerr=importance_df["Importance Std"]
)

plt.gca().invert_yaxis()
plt.xlabel("Mean Decrease in Macro F1-score")
plt.title("Permutation Feature Importance — RBF SVM")
plt.tight_layout()
plt.show()
````

### Feature Importance Analysis

Permutation importance was used to investigate how individual features influence the predictive performance of the optimized RBF SVM.

For each feature, its values were randomly shuffled and the resulting change in Macro F1-score was measured.

A larger positive decrease in performance indicates that the model relies more strongly on that feature.

The analysis was performed on the training set to keep the held-out test set reserved for final evaluation.

Permutation importance measures model reliance rather than causal influence. Results may also be affected by correlations between features.
